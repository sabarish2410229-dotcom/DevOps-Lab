## Importing Packages

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np


from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

from xgboost import XGBClassifier
from imblearn.combine import SMOTETomek

import mlflow
import mlflow.sklearn
import mlflow.xgboost
```

## Load teh Data

```python
import pandas as pd

# Load dataset
data = pd.read_csv(r"C:\Users\Ridhi Rajesh\Downloads\HousingData.csv")

# (Optional) Check for missing values
print(data.isnull().sum())

# Handle missing values
data = data.fillna(data.median(numeric_only=True))

# Features and target
X = data.drop("MEDV", axis=1)
y = data["MEDV"]
```

```python
## print("Shape:", X.shape)
##print("Target:", data.target_names)
##print("Class Dst:", np.unique(y, return_counts=True))
print("Shape:", X.shape)

print("\nColumns:")
print(data.columns)

print("\nTarget Column: MEDV")

print("\nFirst 5 rows:")
print(data.head())

print("\nMissing Values:")
print(data.isnull().sum())

print("\nTarget (MEDV) Statistics:")
print(y.describe())
```

## Train Ready

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)
```

## Diff Exp Models

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create model
lin_reg = LinearRegression()

# Train model
lin_reg.fit(X_train, y_train)

# Predict
y_pred = lin_reg.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create model
rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

# Train
rf.fit(X_train, y_train)

# Predict
y_pred = rf.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create model
xgb = XGBRegressor(
    objective="reg:squarederror",
    n_estimators=100,
    random_state=42
)

# Train model
xgb.fit(X_train, y_train)

# Predict
y_pred = xgb.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

## Prepare for ML Flow

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

models = [

    (
        "Linear Regression",
        LinearRegression(),
        (X_train, y_train),
        (X_test, y_test)
    ),

    (
        "Random Forest",
        RandomForestRegressor(
            n_estimators=100,
            max_depth=5,
            random_state=42
        ),
        (X_train, y_train),
        (X_test, y_test)
    ),

    (
        "XGBoost",
        XGBRegressor(
            objective="reg:squarederror",
            n_estimators=100,
            random_state=42
        ),
        (X_train, y_train),
        (X_test, y_test)
    )

]
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

reports = []

for model_name, model, train_set, test_set in models:

    X_tr, y_tr = train_set
    X_te, y_te = test_set

    model.fit(X_tr, y_tr)

    predictions = model.predict(X_te)

    mae = mean_absolute_error(y_te, predictions)
    mse = mean_squared_error(y_te, predictions)
    rmse = mse ** 0.5
    r2 = r2_score(y_te, predictions)

    reports.append({
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    })
```

```python
reports
```

## Exp Tracking ML Flow Local

```python
mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("Boston Housing Regression PBLM 1")
```

```python
for i, (model_name, model, _, _) in enumerate(models):

    report = reports[i]

    with mlflow.start_run(run_name=model_name):

        # Log model name
        mlflow.log_param("Model", model_name)

        # Log model parameters
        if hasattr(model, "get_params"):
            mlflow.log_params(model.get_params())

        # Log regression metrics
        mlflow.log_metric("MAE", report["MAE"])
        mlflow.log_metric("MSE", report["MSE"])
        mlflow.log_metric("RMSE", report["RMSE"])
        mlflow.log_metric("R2", report["R2"])

        # Log model
        if "XGBoost" in model_name:
            mlflow.xgboost.log_model(
                model,
                artifact_path="model"
            )
        else:
            mlflow.sklearn.log_model(
                model,
                artifact_path="model"
            )
```

## Reg teh Best Model

```python
best_index = np.argmax(
    [report["R2"] for report in reports]
)

best_model_name, best_model, _, _ = models[best_index]

best_report = reports[best_index]

print("Best Model:", best_model_name)
print("R² Score:", best_report["R2"])
print("RMSE:", best_report["RMSE"])
print("MAE:", best_report["MAE"])
print("MSE:", best_report["MSE"])
```

```python
with mlflow.start_run(
    run_name=f"Champion_{best_model_name}"
) as run:

    # Parameters
    mlflow.log_param(
        "Model",
        best_model_name
    )

    mlflow.log_param(
        "Selection_Metric",
        "R2 Score"
    )

    mlflow.log_param(
        "Dataset",
        "Boston Housing Dataset"
    )

    mlflow.log_params(
        best_model.get_params()
    )

    # Regression Metrics
    mlflow.log_metric(
        "R2",
        best_report["R2"]
    )

    mlflow.log_metric(
        "MAE",
        best_report["MAE"]
    )

    mlflow.log_metric(
        "MSE",
        best_report["MSE"]
    )

    mlflow.log_metric(
        "RMSE",
        best_report["RMSE"]
    )

    # Tags
    mlflow.set_tag(
        "Model_Type",
        best_model_name
    )

    mlflow.set_tag(
        "Stage",
        "Candidate"
    )

    mlflow.set_tag(
        "Task",
        "Regression"
    )

    if "XGBoost" in best_model_name:

        model_info = mlflow.xgboost.log_model(
            best_model,
            artifact_path="model",
            registered_model_name="Boston_Housing_Best_Model"
        )

    else:

        model_info = mlflow.sklearn.log_model(
            best_model,
            artifact_path="model",
            registered_model_name="Boston_Housing_Best_Model"
        )

    run_id = run.info.run_id
    model_uri = model_info.model_uri

print("Run ID      :", run_id)
print("Model URI   :", model_uri)
print("Model Name  :", "Boston_Housing_Best_Model")
```

```python
import mlflow
from mlflow.tracking import MlflowClient

mlflow.set_tracking_uri(
    "http://127.0.0.1:5000"
)

client = MlflowClient()
```

```python
model_name = "Boston_Housing_Best_Model"

latest_version = client.get_latest_versions(
    model_name
)[0]

print("Model Name:", latest_version.name)
print("Version:", latest_version.version)
print("Stage:", latest_version.current_stage)
```

```python
model_uri = f"models:/{model_name}/{latest_version.version}"

model = mlflow.pyfunc.load_model(
    model_uri
)

print("Model loaded successfully")
```

```python
predictions = model.predict(
    X_test
)

print(predictions[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = mse ** 0.5
r2 = r2_score(y_test, predictions)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
client.update_model_version(
    name=model_name,
    version=latest_version.version,
    description="""
    Champion model for Boston Housing Price Prediction.

    Tested successfully before production deployment.

    Dataset:
    Boston Housing Dataset (HousingData.csv)

    Task:
    Regression

    Selection Metric:
    R² Score
    """
)
```

```python
client.transition_model_version_stage(
    name=model_name,
    version=latest_version.version,
    stage="Production"
)
```
